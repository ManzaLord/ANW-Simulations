```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
#Inicia las clases
beta_profiles = B_profiles()
plot = plotter(output_dir="figures/wannier")

#Parametros basicos de las guias de onda
params = {
    "L": 50,
    "N": 51,
    "C0": 250,
    "g": 70,
    "points": 500,
    "deltaB": 20,
    "B_s": 7.85e6, #Propagation constant of signal photons
}
params["alpha"] = np.sqrt(0.0005 * params["N"])

#Configuracion del pump (caso no lineal)
pos = int((params["N"] - 1) / 2)
pump_nl = np.zeros(params["N"])
pump_nl[pos] = 1

#Configuracion del pump (caso lineal)
pos_l1 = pos
pos_l2 = pos
pump_l = np.zeros([params["N"],params["N"]])
pump_l[pos_l1][pos_l2] = 1
pump_l[pos_l2][pos_l1] = 1
```

```python
deltaB = np.array([-0.2, 0, 0.2])

wg_l = []
wg_nl = []

for i in range(len(deltaB)):
    
    params["deltaB"] = deltaB[i]
    
    wg = waveguide(params,B_signal=B_profiles.wannier_stark_beta)
    wg.set_pump(pump_nl)
    wg.simulate(calc_schmidt=True)
    wg_nl.append(wg)

    wg = waveguide(params,B_signal=B_profiles.wannier_stark_beta)
    wg.set_pump(pump_l)
    wg.linear(calc_schmidt=True)
    wg_l.append(wg)
```

```python
plot.plot_eigen(wg_list=wg_nl, var=deltaB/params["C0"], var_name='$\\Delta\\beta/C_0$' , wg_number=25, save=True, case="Wannier-Stark")
```

```python
plot.plot_propagation(wg_list=wg_nl, case="Nonlineal", iter_item=deltaB/params["C0"], iter_name=r'\Delta\beta/C_0', save=True)
plot.plot_propagation(wg_list=wg_l, case="Lineal", iter_item=deltaB/params["C0"], iter_name=r'\Delta\beta/C_0', save=True)
```

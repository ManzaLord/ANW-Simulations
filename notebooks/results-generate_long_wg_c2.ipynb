```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
#Inicia las clases
beta_profiles = B_profiles()

#Parametros estandar
params = {
    "C0": 250,
    "g": 70,
    "points": 500,
    "deltaB": 20,
    "B_s": 7.85e6,
}                

params["L"] = 20
params["N"] = 100

params["alpha"] = np.sqrt(0.0005 * params["N"])
```

# Caso 2

```python
#Configuracion del pump (caso no lineal)
pos = int((params["N"] - 1) / 2)
pump_nl2 = np.zeros(params["N"])
pump_nl2[pos] = 1
pump_nl2[pos+1] = 1
```

## Desorden off-diagonal

```python
#Niveles de desorden
kappa_B = [0]
kappa_C = [0, 0.25, 0.5, 0.75]

wg_offdis2 = waveguide(params,B_signal=B_profiles.uniform_beta)#, wg_type="Circular")
wg_offdis2.set_pump(pump_nl2)
wg_offdis2.simulate()
wg_offdis2.disorder(kappa_B,kappa_C, phase_matching=True)#,calc_schmidt=True)

wg_offdis2.save(f"C2_offdiagonal_N{params["N"]}_L{params["L"]}")
```

## Desorden diagonal

```python
#Niveles de desorden
kappa_C = [0]
kappa_B = [0, 0.25, 0.5, 0.75]

wg_dis2 = waveguide(params,B_signal=B_profiles.uniform_beta)#, wg_type="Circular")
wg_dis2.set_pump(pump_nl2)
wg_dis2.simulate()
wg_dis2.disorder(kappa_B,kappa_C, phase_matching=True)#,calc_schmidt=True)

wg_dis2.save(f"C2_diagonal_N{params["N"]}_L{params["L"]}")
```

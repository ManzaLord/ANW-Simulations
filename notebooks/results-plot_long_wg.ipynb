```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
wg_offdis1 = waveguide.load(name="C1_offdiagonal_N100_L20")
wg_offdis2 = waveguide.load(name="C2_offdiagonal_N100_L20")
wg_offdis3 = waveguide.load(name="C3_offdiagonal_N100_L20")

wg_dis1 = waveguide.load(name="C1_diagonal_N100_L20")
wg_dis2 = waveguide.load(name="C2_diagonal_N100_L20")
wg_dis3 = waveguide.load(name="C3_diagonal_N100_L20")

kappa_C = [0, 0.25, 0.5, 0.75]
kappa_B = [0, 0.25, 0.5, 0.75]
```

# Caso 1

```python
#Configura el plotter
plot = plotter(output_dir="figures/results/case1")
```

## Propagación

```python
plot.plot_propagation(wg_offdis1, case= "C1_N100", save=True)
```

## Desorden off-diagonal

```python
plot.plot_disorder_corr(wg_offdis1,iter_item=kappa_C, iter_name="\kappa_C", corr_type="Q", save=True, case="C1_offdiag")
```

## Desorden diagonal

```python
plot.plot_disorder_corr(wg_dis1,iter_item=kappa_B, iter_name=r"\kappa_\beta", corr_type="Q", save=True, case="C1_diag")
```

```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
#Inicia las clases
beta_profiles = B_profiles()

#Parametros estandar
params = {
    "N": 10,
    "C0": 250,
    "g": 70,
    "points": 500,
    "deltaB": 20,
    "B_s": 7.85e6,
}                
params["alpha"] = np.sqrt(0.0005 * params["N"])
```

# Caso 1

```python
#Configura el plotter
plot = plotter(output_dir="figures/results/case1")

#Configuracion del pump (caso no lineal)
pos = int((params["N"] - 1) / 2)
pump_nl1 = np.zeros(params["N"])
pump_nl1[pos] = 1
```

## Desorden off-diagonal

```python
#Niveles de desorden
kappa_B = [0]
kappa_C = [0, 0.25, 0.5, 0.75]

L_vals = [0.5, 1.203, 2]

wg_offdis1 = [None] * len(L_vals)

for i in range(len(L_vals)):
    
    params["L"] = L_vals[i]

    wg_offdis1[i] = waveguide(params,B_signal=B_profiles.uniform_beta)#, wg_type="Circular")
    wg_offdis1[i].set_pump(pump_nl1)
    wg_offdis1[i].simulate()
    wg_offdis1[i].disorder(kappa_B,kappa_C)#,calc_schmidt=True)
```

```python
for i in range(len(L_vals)):
    plot.plot_disorder_corr(wg_offdis1[i],iter_item=kappa_C, iter_name="\kappa_C", corr_type="Q", limits=[2,7], save=True, case="C1_offdiag")
```

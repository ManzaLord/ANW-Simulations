```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
#Inicia las clases
beta_profiles = B_profiles()

#Parametros estandar
params = {
    "C0": 250,
    "g": 70,
    "points": 500,
    "deltaB": 20,
    "B_s": 7.85e6,
}                

params["L"] = 20
params["N"] = 100

params["alpha"] = np.sqrt(0.0005 * params["N"])
```

# Caso 3

```python
#Configuracion del pump (caso no lineal)
pos = int((params["N"] - 1) / 2)
pump_nl3 = np.zeros(params["N"])
pump_nl3[pos] = 1
pump_nl3[pos+1] = -1
```

## Desorden off-diagonal

```python
#Niveles de desorden
kappa_B = [0]
kappa_C = [0, 0.25, 0.5, 0.75]

wg_offdis3 = waveguide(params,B_signal=B_profiles.uniform_beta)#, wg_type="Circular")
wg_offdis3.set_pump(pump_nl3)
wg_offdis3.simulate()
wg_offdis3.disorder(kappa_B,kappa_C)#,calc_schmidt=True)

wg_offdis3.save(f"C3_offdiagonal_N{params["N"]}_L{params["L"]}")
```

## Desorden diagonal

```python
#Niveles de desorden
kappa_C = [0]
kappa_B = [0, 0.25, 0.5, 0.75]

wg_dis3 = waveguide(params,B_signal=B_profiles.uniform_beta)#, wg_type="Circular")
wg_dis3.set_pump(pump_nl3)
wg_dis3.simulate()
wg_dis3.disorder(kappa_B,kappa_C)#,calc_schmidt=True)

wg_dis3.save(f"C3_diagonal_N{params["N"]}_L{params["L"]}")
```

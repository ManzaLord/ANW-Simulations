```python
import sys
import numpy as np
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from waveguide import waveguide, plotter, B_profiles
```

```python
#Inicia las clases
beta_profiles = B_profiles()
plot = plotter(output_dir="figures/short_circular/nonlinear")

#Parametros estandar
params = {
    "L": 50,
    "C0": 250,
    "g": 70,
    "points": 500,
    "deltaB": 20,
    "B_s": 7.85e6,
}                

#Niveles de desorden
kappa_B = [0]
kappa_C = [0, 0.25, 0.5, 0.75]
```

```python
N_vals = [5, 6, 7, 8]

wg_circ_list = [None] * len(N_vals)

for i in range(len(N_vals)):
    
    params["N"] = N_vals[i]
    params["alpha"] = np.sqrt(0.0005 * params["N"])

    #Configuracion del pump (caso no lineal)
    pos = int((params["N"] - 1) / 2)
    pump_nl = np.zeros(params["N"])
    pump_nl[pos] = 1

    wg_circ_list[i] = waveguide(params,B_signal=B_profiles.uniform_beta, wg_type="Circular")
    wg_circ_list[i].set_pump(pump_nl)
    wg_circ_list[i].disorder(kappa_B,kappa_C)#,calc_schmidt=True)
```

```python
for i in range(len(N_vals)):
    title = f"Circular array with {N_vals[i]} waveguides, central pump"
    plot.plot_disorder_propagation(wg_circ_list[i], iter_item=kappa_C, iter_name=r"\kappa_{C}", case=title, zlim=3, save=True)
```

```python
for i in range(len(N_vals)):
    title = f"Circular array with {N_vals[i]} waveguides, central pump"
    plot.plot_disorder_corr(wg_circ_list[i], corr_type="Q", case=title, z_val=5)#, save=True)
    plot.plot_disorder_corr(wg_circ_list[i], corr_type="K", case=title, z_val=5)#, save=True)
```

```python
plot.plot_disorder_data(wg_circ_list, "schmidt", case="Central pump, Circular", ncols=4, save=True)
plot.plot_disorder_data(wg_circ_list, "std", case="Central pump, Circular", ncols=4, save=True)
plot.plot_disorder_data(wg_circ_list, "std_der", case="Central pump, Circular", ncols=4, save=True)
```

```python
for i in range(len(N_vals)):
    title = f"Nonlinear circular array with {N_vals[i]} waveguides, central pump"
    plot.animate_disorder_corr(wg_circ_list[i], corr_type="Q", case=title, save=True)
    plot.animate_disorder_corr(wg_circ_list[i], corr_type="K", case=title, save=True)
```
